```python
# 1. Creating lists (biology examples)
sl_genes = ["POLQ", "BRCA1", "CTNNB1", "TERT", "TP53", "AXIN1"]   # from your manuscript Table 2
liver_expressed = ["POLQ", "BRCA1", "KEAP1", "ARID1A"]
patient_ids = ["TCGA-LIHC-01A", "TCGA-LIHC-02B", "TCGA-LIHC-03C"]

print(len(sl_genes))           # 6
print(sl_genes[0])             # POLQ (index starts at 0)
print(sl_genes[-1])            # last item = AXIN1
```

```python
# 2. Adding / Removing items
sl_genes.append("MDM4")        # add new SL partner
sl_genes.remove("TP53")        # remove one
print(sl_genes)

# 3. Slicing (very useful for splitting train/test)
top_3 = sl_genes[:3]           # first 3
print(top_3)
```

```python
# Example 1: Uppercase all gene names
upper_genes = [gene.upper() for gene in sl_genes]
print(upper_genes)

# Example 2: Filter only genes that appear in SynLethDB (simulated)
known_sl = ["POLQ", "BRCA1", "CTNNB1"]
high_confidence = [gene for gene in sl_genes if gene in known_sl]
print(high_confidence)

# Example 3: Create SL pair labels (exactly like Phase 3)
pairs = [f"{g1}_vs_{g2}" for g1 in sl_genes for g2 in liver_expressed if g1 != g2]
print(len(pairs))               # many pairs
print(pairs[:5])                # first 5
```

```python
# Real project example: Filter liver-expressed genes with expression > 5
expression_values = [8.45, 3.2, 7.8, 1.1, 9.3]   # fake TPM for 5 genes
high_expr_genes = [gene for gene, expr in zip(sl_genes, expression_values) if expr > 5]
print(high_expr_genes)   # → genes you will keep for your model
```

```python
# 1. Lowercase gene names
lower_genes = [gene.lower() for gene in sl_genes]
print(lower_genes)

# 2. Filter tumor suppressors (example)
tumor_suppressors = ["TP53", "BRCA1", "ARID1A"]
filtered_ts = [gene for gene in sl_genes if gene in tumor_suppressors]
print(filtered_ts)

# 3. Remove TP53 using comprehension
no_tp53 = [gene for gene in sl_genes if gene != "TP53"]
print(no_tp53)

# 4. Create gene pairs with underscore
pair_list = [g1 + "_" + g2 for g1 in sl_genes for g2 in liver_expressed]
print(pair_list[:5])

# 5. Add prefix "Gene:"
labeled = [f"Gene:{gene}" for gene in sl_genes]
print(labeled)

#  6. Uppercase gene names
upper_genes = [gene.upper() for gene in sl_genes]
print(upper_genes)

# 7. Length of gene names
gene_lengths = [len(gene) for gene in sl_genes]
print(gene_lengths)

# 8. Combine gene + expression
combined = [f"{gene}:{expr}" for gene, expr in zip(sl_genes, expression_values)]
print(combined)
```
