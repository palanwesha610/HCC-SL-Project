```python
# Gene expression dictionary
gene_expression = {
    "POLQ": 8.45,
    "BRCA1": 6.2,
    "CTNNB1": 9.1,
    "TERT": 7.8
}

# SL scores
gene_sl_score = {
    "POLQ": 0.97,
    "BRCA1": 0.95,
    "AXIN1": 0.92
}

# Access & update
print(gene_expression["POLQ"])
gene_expression["POLQ"] = 8.7

# Safe access
print(gene_expression.get("TP53", "Not found"))

# Loop
for gene, score in gene_sl_score.items():
    if score > 0.9:
        print(f"High-confidence SL: {gene}")
```

```python
# Sets (unique elements)
known_sl_genes = {"POLQ", "BRCA1", "CTNNB1", "POLQ"}
liver_genes = {"POLQ", "BRCA1", "KEAP1", "ARID1A"}

# Membership check
print("POLQ" in liver_genes)

# Set operations
common = known_sl_genes & liver_genes
all_candidates = known_sl_genes | liver_genes

print(common)
print(all_candidates)
```

```python
sl_candidates = {
    ("POLQ", "BRCA1"),
    ("CTNNB1", "AXIN1"),
    ("TERT", "TERF2")
}

high_confidence_set = {pair for pair in sl_candidates if pair[0] in liver_genes}
print(high_confidence_set)
```
