```python
# 1. Creating strings (biology examples)
gene_a = "POLQ"                    # top SL gene from your manuscript
gene_b = 'BRCA1'                   # single quotes also work
patient_id = "TCGA-LIHC-01A"       # real TCGA barcode
sl_pair = gene_a + "-" + gene_b    # concatenation
print("SL pair:", sl_pair)         # POLQ-BRCA1

# 2. f-strings (modern & clean — you will use this every day)
score = 0.97
print(f"{gene_a}-{gene_b} predicted SL score = {score:.3f}")
# Output: POLQ-BRCA1 predicted SL score = 0.970
```

```python
# 3. Useful String Methods (memorize these 8 — they appear in every bioinformatics script)

text = "  TCGA-LIHC-01A   "          # messy TCGA ID with spaces

print(text.strip())                 # removes spaces → "TCGA-LIHC-01A"
print(text.upper())                 # "  TCGA-LIHC-01A   "
print(text.lower())                 # "  tcga-lihc-01a   "
print("POLQ".replace("Q", "Q-inhibitor"))   # replace parts
print(patient_id.split("-"))        # ['TCGA', 'LIHC', '01A'] ← super useful!
print("BRCA1" in patient_id)        # False
print(len(gene_a))                  # 4 (length of gene name)
print(patient_id.startswith("TCGA"))# True
print(patient_id.find("LIHC"))      # 5 (position of "LIHC")
```

```python
# 4. Real project example: Building gene-pair feature label
gene1 = "POLQ"
gene2 = "BRCA1"
pair_label = f"{gene1}_vs_{gene2}_SL"
print(pair_label)                   # POLQ_vs_BRCA1_SL

# 5. Checking gene names (important for filtering liver-expressed genes)
liver_genes = ["POLQ", "CTNNB1", "TERT", "TP53"]
query = "polq"
if query.upper() in [g.upper() for g in liver_genes]:
    print("Gene found in liver set!")
```
