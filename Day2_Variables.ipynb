```python
# 1. String (text) - for gene names, patient IDs
gene_name = "POLQ"
print(type(gene_name))          # <class 'str'>

# 2. Integer (whole numbers) - patient counts, gene IDs
tcga_patients = 370
print(type(tcga_patients))      # <class 'int'>

# 3. Float (decimal numbers) - expression, fold-change, SL scores
polq_expression = 8.45
sl_score = 0.97
print(type(polq_expression))    # <class 'float'>

# 4. Boolean (True/False) - mutation status, HRD flag, SL label
is_hrd = True
has_drug = False
print(type(is_hrd))             # <class 'bool'>

# 5. NoneType (empty value) - useful for missing data
missing_value = None
print(type(missing_value))
```

```python
# Arithmetic (for log2FC, correlations)
normal = 4.2
tumor = 8.45
log2fc = tumor - normal
print("POLQ log2FC =", log2fc)

# Comparison (for filtering genes)
print(tumor > 5.0)          # True
print(sl_score >= 0.9)      # True

# Logical (for SL label assignment)
both_druggable = True and has_drug
print("Both druggable?", both_druggable)

# Membership (later for pathway overlap)
liver_genes = ["POLQ", "BRCA1", "CTNNB1"]
print("POLQ" in liver_genes)   # True
```
